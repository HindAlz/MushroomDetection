```python

import warnings
warnings.filterwarnings("ignore")


import numpy as np
import pandas as pd


from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OrdinalEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split, StratifiedKFold, GridSearchCV
from sklearn.metrics import (
accuracy_score, precision_score, recall_score, f1_score,
roc_auc_score, classification_report, confusion_matrix
)
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neural_network import MLPClassifier
# Optional XGBoost (CPU hist)
try:
    from xgboost import XGBClassifier
    HAVE_XGB = True
except Exception:
    HAVE_XGB = False

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

CSV_PATH = "secondary_data.csv"  # change if needed
SEP = ";"                    # your file uses semicolon
```

```python
#load data
df = pd.read_csv(CSV_PATH, sep=SEP)
```

```python
# ==================================
# Cell 2: Load data & static cleanup
# ==================================
df = pd.read_csv(CSV_PATH, sep=SEP)

STATIC_DROP = ["veil-type", "spore-print-color", "veil-color", "stem-root"]
for c in STATIC_DROP:
    if c in df.columns:
        df = df.drop(columns=c)
```

```python

# ==========================================
# Cell 3: Target/feature split (no encoding)
# ==========================================
y_raw = df["class"].astype(str)
y = y_raw.map({"e": 0, "p": 1}).fillna(y_raw).astype(int)
X = df.drop(columns="class")

# ======================================
# Cell 4: Train/Test split (stratified)
# ======================================
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=RANDOM_STATE, stratify=y
)
```

```python

# ==========================================================
# Cell 5: Preprocessing built from TRAIN columns (no leak)
# ==========================================================
cat_cols = X_train.select_dtypes(include=["object", "category", "string"]).columns.tolist()
num_cols = X_train.select_dtypes(include=["number"]).columns.tolist()

pre_cat = Pipeline([
    ("impute", SimpleImputer(strategy="most_frequent")),
    ("encode", OrdinalEncoder(handle_unknown="use_encoded_value", unknown_value=-1)),
])

pre_num = Pipeline([
    ("impute", SimpleImputer(strategy="median")),
    ("scale", StandardScaler(with_mean=True, with_std=True)),
])

preprocessor = ColumnTransformer(
    transformers=[
        ("cat", pre_cat, cat_cols),
        ("num", pre_num, num_cols),
    ],
    remainder="drop",
)
```

```python

# ==========================================
# Cell 6: Define models & hyperparam grids
# ==========================================
models = []

# Logistic Regression
lr_pipe = Pipeline([
    ("prep", preprocessor),
    ("clf", LogisticRegression(max_iter=500, solver="liblinear", random_state=RANDOM_STATE)),
])
lr_grid = {"clf__C": [0.01, 0.1, 1.0, 10.0]}
models.append(("Logistic Regression", lr_pipe, lr_grid))

# SVM
svm_pipe = Pipeline([
    ("prep", preprocessor),
    ("clf", SVC(kernel="rbf", probability=True, random_state=RANDOM_STATE)),
])
svm_grid = {"clf__C": [0.1, 1, 10], "clf__gamma": ["scale", "auto"]}
models.append(("SVM (RBF)", svm_pipe, svm_grid))

# Neural Network (MLP)
mlp_pipe = Pipeline([
    ("prep", preprocessor),
    ("clf", MLPClassifier(hidden_layer_sizes=(64, 32), max_iter=500, random_state=RANDOM_STATE)),
])
mlp_grid = {"clf__alpha": [0.0001, 0.001, 0.01]}
models.append(("Neural Network (MLP)", mlp_pipe, mlp_grid))

# Random Forest
rf_pipe = Pipeline([
    ("prep", preprocessor),
    ("clf", RandomForestClassifier(random_state=RANDOM_STATE, n_jobs=-1)),
])
rf_grid = {
    "clf__n_estimators": [100, 200],
    "clf__max_depth": [None, 15, 30],
}
models.append(("Random Forest", rf_pipe, rf_grid))

# XGBoost (optional)
if HAVE_XGB:
    xgb_pipe = Pipeline([
        ("prep", preprocessor),
        ("clf", XGBClassifier(
            tree_method="hist", predictor="auto", eval_metric="logloss",
            random_state=RANDOM_STATE, subsample=0.8, colsample_bytree=0.8,
            n_estimators=300, learning_rate=0.05, max_depth=6,
        )),
    ])
    xgb_grid = {
        "clf__max_depth": [4, 6, 8],
        "clf__subsample": [0.7, 0.8, 1.0],
    }
    models.append(("XGBoost", xgb_pipe, xgb_grid))
```

```python

# =====================================
# Cell 7: Tuning (5-fold stratified CV)
# =====================================
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

results = []
for name, pipe, grid in models:
    print(f"\nTraining {name}...")
    gs = GridSearchCV(pipe, grid, scoring="accuracy", cv=cv, n_jobs=-1, refit=True)
    gs.fit(X_train, y_train)

    est = gs.best_estimator_
    y_pred = est.predict(X_test)

    if hasattr(est, "predict_proba"):
        y_proba = est.predict_proba(X_test)[:, 1]
    elif hasattr(est, "decision_function"):
        s = est.decision_function(X_test)
        s_min, s_max = s.min(), s.max()
        y_proba = (s - s_min) / (s_max - s_min + 1e-12)
    else:
        y_proba = y_pred.astype(float)

    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred)
    rec  = recall_score(y_test, y_pred)
    f1   = f1_score(y_test, y_pred)
    roc  = roc_auc_score(y_test, y_proba)

    results.append({
        "Model": name,
        "Accuracy": round(acc, 4),
        "Precision": round(prec, 4),
        "Recall": round(rec, 4),
        "F1": round(f1, 4),
        "ROC_AUC": round(roc, 4),
        "BestParams": gs.best_params_,
        "Estimator": est,
    })

```

```python
# ============================================
# Cell 8: Results summary & best model report
# ============================================
import pandas as pd
from IPython.display import display
import numpy as np

# Sort by accuracy descending
results_df = pd.DataFrame(results).sort_values(by="Accuracy", ascending=False)

# Build a compact summary (hide heavy estimator objects)
summary_cols = ["Model", "Accuracy", "Precision", "Recall", "F1", "ROC_AUC", "BestParams"]
summary = results_df[summary_cols].copy()

# Round metrics nicely
for col in ["Accuracy", "Precision", "Recall", "F1", "ROC_AUC"]:
    summary[col] = summary[col].map(lambda x: f"{x:.4f}")

# Shorten parameter dict display
def short_params(d):
    if isinstance(d, dict):
        items = [f"{k.split('__')[-1]}={v}" for k, v in d.items()]
        return ", ".join(items)
    return str(d)

summary["BestParams"] = summary["BestParams"].apply(short_params)

print("\n=== Model Comparison (Leakage-safe) ===")
display(summary.style.hide(axis="index").set_caption("Model Performance Summary"))

# Pick best model
best_row = results_df.iloc[0]
best_name = best_row["Model"]
best_estimator = best_row["Estimator"]

print(f"\n=== Best Model Detailed Report ===")
print(f"Model: {best_name}")
print(f"Best Params: {short_params(best_row['BestParams'])}")

# Predict and evaluate
best_pred = best_estimator.predict(X_test)
print("\nClassification Report:\n")
print(classification_report(y_test, best_pred, digits=4))

cm = confusion_matrix(y_test, best_pred)
cm_df = pd.DataFrame(cm, index=["True: Edible (0)", "True: Poisonous (1)"],
                     columns=["Pred: Edible (0)", "Pred: Poisonous (1)"])
print("\nConfusion Matrix:")
display(cm_df)
```

```python
# ============================================
# Cell 9: Feature selection — L1 (Logistic Regression)
# ============================================
from sklearn.feature_selection import SelectFromModel

# Build a pipeline: preprocessing -> L1-based selector -> RandomForest
l1_selector = SelectFromModel(
    LogisticRegression(penalty="l1", solver="liblinear", C=0.1, random_state=RANDOM_STATE)
)

fs_pipe = Pipeline([
    ("prep", preprocessor),
    ("select", l1_selector),
    ("clf", RandomForestClassifier(random_state=RANDOM_STATE, n_jobs=-1))
])

# Train/test split remains the same
fs_pipe.fit(X_train, y_train)

# Get selected features (after preprocessing)
mask = fs_pipe.named_steps["select"].get_support()
feature_names = np.array(cat_cols + num_cols)[mask]
print(f"Selected {mask.sum()} features out of {len(mask)}:")
print(feature_names)

# Evaluate
y_pred = fs_pipe.predict(X_test)
print("\nAccuracy after feature selection:", accuracy_score(y_test, y_pred))
```

```python
# ============================================
# Cell 10: Feature selection — ANOVA F-test
# ============================================
from sklearn.feature_selection import SelectKBest, f_classif

anova_pipe = Pipeline([
    ("prep", preprocessor),
    ("select", SelectKBest(score_func=f_classif, k=10)),  # choose k
    ("clf", RandomForestClassifier(random_state=RANDOM_STATE, n_jobs=-1))
])

anova_pipe.fit(X_train, y_train)

y_pred = anova_pipe.predict(X_test)
print("\nAccuracy after ANOVA feature selection:", accuracy_score(y_test, y_pred))
```

```python
# ============================================
# Cell 11: Feature selection — Tree-based importance
# ============================================
from sklearn.feature_selection import SelectFromModel

tree_selector = SelectFromModel(
    RandomForestClassifier(n_estimators=200, random_state=RANDOM_STATE, n_jobs=-1),
    threshold="median"  # keep top 50% features
)

tree_pipe = Pipeline([
    ("prep", preprocessor),
    ("select", tree_selector),
    ("clf", RandomForestClassifier(random_state=RANDOM_STATE, n_jobs=-1))
])

tree_pipe.fit(X_train, y_train)
mask = tree_pipe.named_steps["select"].get_support()
feature_names = np.array(cat_cols + num_cols)[mask]
print(f"Selected {mask.sum()} features (≥ median importance):")
print(feature_names)
```

```python

# ============================================
# Cell 9: SVM-only — baseline & feature selection (leakage-safe)
# ============================================
# This cell assumes you already ran the previous cells to define:
# - X_train, X_test, y_train, y_test
# - preprocessor, cat_cols, num_cols

from sklearn.pipeline import Pipeline
from sklearn.svm import SVC
from sklearn.feature_selection import SelectFromModel, SelectKBest, f_classif
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import StratifiedKFold, GridSearchCV
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, classification_report, confusion_matrix
)

svm_results = []
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

# --------------------------------------------
# 9.1 Baseline SVM (RBF) — with global scaling
# --------------------------------------------
# We add a StandardScaler AFTER the ColumnTransformer so both encoded cats
# and numeric features are on a comparable scale for the RBF kernel.
svm_base = Pipeline([
    ("prep", preprocessor),
    ("scale", StandardScaler(with_mean=True, with_std=True)),
    ("clf", SVC(kernel="rbf", probability=True, random_state=RANDOM_STATE)),
])

svm_base_grid = {
    "clf__C": [0.1, 1, 10],
    "clf__gamma": ["scale", "auto"],
}

svm_base_gs = GridSearchCV(svm_base, svm_base_grid, scoring="accuracy", cv=cv, n_jobs=-1, refit=True)
svm_base_gs.fit(X_train, y_train)

svm_base_est = svm_base_gs.best_estimator_
svm_base_pred = svm_base_est.predict(X_test)
svm_base_proba = (
    svm_base_est.predict_proba(X_test)[:,1]
    if hasattr(svm_base_est, "predict_proba") else svm_base_pred.astype(float)
)

svm_results.append({
    "Variant": "SVM (baseline)",
    "BestParams": svm_base_gs.best_params_,
    "Accuracy": round(accuracy_score(y_test, svm_base_pred), 4),
    "Precision": round(precision_score(y_test, svm_base_pred), 4),
    "Recall": round(recall_score(y_test, svm_base_pred), 4),
    "F1": round(f1_score(y_test, svm_base_pred), 4),
    "ROC_AUC": round(roc_auc_score(y_test, svm_base_proba), 4),
    "Estimator": svm_base_est,
})

```

```python

# -----------------------------------------------------
# 9.2 SVM with L1-based selection (LogReg + SelectFromModel)
# -----------------------------------------------------
# L1 (sparse) LR selects features → then scaled → SVM (RBF)
svm_l1 = Pipeline([
    ("prep", preprocessor),
    ("scale", StandardScaler(with_mean=True, with_std=True)),
    ("select", SelectFromModel(LogisticRegression(penalty="l1", solver="liblinear", C=0.1, random_state=RANDOM_STATE))),
    ("clf", SVC(kernel="rbf", probability=True, random_state=RANDOM_STATE)),
])

svm_l1_grid = {
    "select__estimator__C": [0.05, 0.1, 0.2, 0.5, 1.0],
    "clf__C": [0.1, 1, 10],
    "clf__gamma": ["scale", "auto"],
}

svm_l1_gs = GridSearchCV(svm_l1, svm_l1_grid, scoring="accuracy", cv=cv, n_jobs=-1, refit=True)
svm_l1_gs.fit(X_train, y_train)

svm_l1_est = svm_l1_gs.best_estimator_
svm_l1_pred = svm_l1_est.predict(X_test)
svm_l1_proba = (
    svm_l1_est.predict_proba(X_test)[:,1]
    if hasattr(svm_l1_est, "predict_proba") else svm_l1_pred.astype(float)
)

# Recover selected feature names (original order cat+num)
try:
    l1_mask = svm_l1_est.named_steps["select"].get_support()
    l1_feature_names = np.array(cat_cols + num_cols)[l1_mask]
except Exception:
    l1_feature_names = []

svm_results.append({
    "Variant": "SVM + L1 feature selection",
    "BestParams": svm_l1_gs.best_params_,
    "Accuracy": round(accuracy_score(y_test, svm_l1_pred), 4),
    "Precision": round(precision_score(y_test, svm_l1_pred), 4),
    "Recall": round(recall_score(y_test, svm_l1_pred), 4),
    "F1": round(f1_score(y_test, svm_l1_pred), 4),
    "ROC_AUC": round(roc_auc_score(y_test, svm_l1_proba), 4),
    "Estimator": svm_l1_est,
    "SelectedFeatures": l1_feature_names,
})
```

```python

# ---------------------------------------------
# 9.3 SVM with ANOVA (SelectKBest) feature filter
# ---------------------------------------------
svm_anova = Pipeline([
    ("prep", preprocessor),
    ("scale", StandardScaler(with_mean=True, with_std=True)),
    ("select", SelectKBest(score_func=f_classif, k=10)),  # adjust k if desired
    ("clf", SVC(kernel="rbf", probability=True, random_state=RANDOM_STATE)),
])

svm_anova_grid = {
    "select__k": [5, 10, 15],
    "clf__C": [0.1, 1, 10],
    "clf__gamma": ["scale", "auto"],
}

svm_anova_gs = GridSearchCV(svm_anova, svm_anova_grid, scoring="accuracy", cv=cv, n_jobs=-1, refit=True)
svm_anova_gs.fit(X_train, y_train)

svm_anova_est = svm_anova_gs.best_estimator_
svm_anova_pred = svm_anova_est.predict(X_test)
svm_anova_proba = (
    svm_anova_est.predict_proba(X_test)[:,1]
    if hasattr(svm_anova_est, "predict_proba") else svm_anova_pred.astype(float)
)

# Attempt to map KBest mask to original names
try:
    anova_mask = svm_anova_est.named_steps["select"].get_support()
    anova_feature_names = np.array(cat_cols + num_cols)[anova_mask]
except Exception:
    anova_feature_names = []

svm_results.append({
    "Variant": "SVM + ANOVA (KBest)",
    "BestParams": svm_anova_gs.best_params_,
    "Accuracy": round(accuracy_score(y_test, svm_anova_pred), 4),
    "Precision": round(precision_score(y_test, svm_anova_pred), 4),
    "Recall": round(recall_score(y_test, svm_anova_pred), 4),
    "F1": round(f1_score(y_test, svm_anova_pred), 4),
    "ROC_AUC": round(roc_auc_score(y_test, svm_anova_proba), 4),
    "Estimator": svm_anova_est,
    "SelectedFeatures": anova_feature_names,
})
```

```python

# ---------------------------------------------
# 9.4 SVM Results — summary & best variant
# ---------------------------------------------
svm_results_df = pd.DataFrame(svm_results).sort_values(by="Accuracy", ascending=False)

from IPython.display import display, HTML

# Pretty summary table
svm_summary_cols = ["Variant", "Accuracy", "Precision", "Recall", "F1", "ROC_AUC", "BestParams"]
svm_summary = svm_results_df[svm_summary_cols].copy()
for c in ["Accuracy", "Precision", "Recall", "F1", "ROC_AUC"]:
    svm_summary[c] = svm_summary[c].map(lambda x: f"{x:.4f}")

def _short(d):
    return ", ".join(f"{k.split('__')[-1]}={v}" for k, v in d.items()) if isinstance(d, dict) else str(d)
svm_summary["BestParams"] = svm_summary["BestParams"].map(_short)

print("
=== SVM Variants Comparison (Leakage-safe) ===")
display(HTML(svm_summary.to_html(index=False)))

# Best SVM variant
svm_best_row = svm_results_df.iloc[0]
svm_best_name = svm_best_row["Variant"]
svm_best_est  = svm_best_row["Estimator"]

print(f"
=== Best SVM Variant: {svm_best_name} ===")
print(f"Best Params: {_short(svm_best_row['BestParams'])}")

svm_best_pred = svm_best_est.predict(X_test)
print("Classification Report (Best SVM):")
print(classification_report(y_test, svm_best_pred, digits=4))

cm = confusion_matrix(y_test, svm_best_pred)
cm_df = pd.DataFrame(cm, index=["true_e(0)", "true_p(1)"], columns=["pred_e(0)", "pred_p(1)"])
print("Confusion Matrix (Best SVM):")
display(cm_df)

# Show selected features for the best variant if available
sel = svm_best_row.get("SelectedFeatures", [])
if isinstance(sel, (list, np.ndarray)) and len(sel) > 0:
    print("Selected features (best SVM):")
    for f in sel:
        print(" -", f)
```
