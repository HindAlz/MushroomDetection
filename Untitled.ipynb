```python
# ================================
# Feature selection + model tuning
# ================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, RandomizedSearchCV
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.metrics import confusion_matrix, classification_report
from sklearn.linear_model import LogisticRegression
from sklearn.feature_selection import SelectKBest, f_classif, SelectFromModel
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from scipy.stats import randint
from sklearn.preprocessing import LabelEncoder
# -------------------------
# 0) Target & features
# -------------------------
df = df.copy()

le = LabelEncoder()
for col in df.select_dtypes(include='object').columns:
    df[col] = le.fit_transform(df[col])
    
# Handle class encoding: accept 'e'/'p' or 0/1
if df['class'].dtype == 'O':
    y = df['class'].map({'e': 0, 'p': 1})
else:
    y = df['class'].astype(int)

X = df.drop(columns='class')

# Train/test split (same 80/20 style)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# -----------------------------------
# Helper for perf formatting (like theirs)
# -----------------------------------
def perf_metrics(y_true, y_pred):
    return {
        "accuracy": round(accuracy_score(y_true, y_pred), 4),
        "f1": round(f1_score(y_true, y_pred), 4),
        "precision": round(precision_score(y_true, y_pred), 4),
        "recall": round(recall_score(y_true, y_pred), 4),
    }

# =========================
# STEP 2: Variable Selection
# =========================

# -------- A) L1-Logistic (“Lasso”-style) selection (LEAKAGE-SAFE)
# Fit on TRAIN ONLY, no scaling needed for tree models later
# For sparse selection use C as a knob (lower C => stronger sparsity)
l1_selector = SelectFromModel(
    LogisticRegression(penalty='l1', solver='liblinear', C=1.0, random_state=42)
)
l1_selector.fit(X_train, y_train)
l1_mask = l1_selector.get_support()
l1_selected_features = X_train.columns[l1_mask]

print("\n[L1 selection] selected features:")
print(l1_selected_features.tolist())

# Project train/test to selected features
X_train_l1 = X_train[l1_selected_features]
X_test_l1  = X_test[l1_selected_features]

# -------- B) KNN “selection” shown in the other notebook actually used ANOVA F-test
# (We replicate it as a filter selector; fit on TRAIN ONLY)
kbest = SelectKBest(score_func=f_classif, k=min(5, X_train.shape[1]))
kbest.fit(X_train, y_train)
kbest_mask = kbest.get_support()
kbest_selected_features = X_train.columns[kbest_mask]

print("\n[ANOVA F-test (k=5)] selected features:")
print(kbest_selected_features.tolist())

# -------- C) Correlation filter (unsupervised—uses X only)
corr = X_train.corr(numeric_only=True)
threshold = 0.7
to_drop = set()
cols = corr.columns
for i in range(len(cols)):
    for j in range(i):
        if abs(corr.iloc[i, j]) > threshold:
            to_drop.add(cols[i])

X_train_corr = X_train.drop(columns=list(to_drop))
X_test_corr = X_test[X_train_corr.columns]

print("\n[Correlation filter | threshold=0.7] kept features:")
print(X_train_corr.columns.tolist())

# =========================
# STEP 3: Models + Tuning
# We’ll use the L1-selected set (like the other notebook picks one path)
# =========================

# ---- Baselines (before tuning) on L1-selected features
rf = RandomForestClassifier(random_state=42)
knn = KNeighborsClassifier()
dt  = DecisionTreeClassifier(max_depth=20, criterion="gini", random_state=42)

rf.fit(X_train_l1, y_train)
knn.fit(X_train_l1, y_train)
dt.fit(X_train_l1, y_train)

rf_pred_b = rf.predict(X_test_l1)
knn_pred_b = knn.predict(X_test_l1)
dt_pred_b = dt.predict(X_test_l1)

acc_rf_b = accuracy_score(y_test, rf_pred_b)
acc_knn_b = accuracy_score(y_test, knn_pred_b)
acc_dt_b = accuracy_score(y_test, dt_pred_b)

print(f"\nBaseline Accuracies (L1-selected):")
print(f"  Random Forest: {acc_rf_b:.4f}")
print(f"  KNN          : {acc_knn_b:.4f}")
print(f"  DecisionTree : {acc_dt_b:.4f}")

# ---- RandomizedSearchCV (same style/params as theirs)

# Random Forest params
param_dist_rf = {
    'n_estimators': [50, 100, 150, 200],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}
rf_rs = RandomizedSearchCV(
    RandomForestClassifier(random_state=42),
    param_distributions=param_dist_rf,
    n_iter=10, cv=5, scoring='accuracy', random_state=42, n_jobs=-1
)
rf_rs.fit(X_train_l1, y_train)
rf_best = rf_rs.best_estimator_
rf_pred_a = rf_best.predict(X_test_l1)
acc_rf_a = accuracy_score(y_test, rf_pred_a)

print("\n[Random Forest] best params:", rf_rs.best_params_)
print(f"[Random Forest] accuracy after tuning: {acc_rf_a:.4f}")

# KNN params
param_dist_knn = {
    'n_neighbors': [3, 5, 7, 9],
    'weights': ['uniform', 'distance'],
    'p': [1, 2]
}
knn_rs = RandomizedSearchCV(
    KNeighborsClassifier(),
    param_distributions=param_dist_knn,
    n_iter=10, cv=5, scoring='accuracy', random_state=42, n_jobs=-1
)
knn_rs.fit(X_train_l1, y_train)
knn_best = knn_rs.best_estimator_
knn_pred_a = knn_best.predict(X_test_l1)
acc_knn_a = accuracy_score(y_test, knn_pred_a)

print("\n[KNN] best params:", knn_rs.best_params_)
print(f"[KNN] accuracy after tuning: {acc_knn_a:.4f}")

# Decision Tree params
param_dist_dt = {
    "max_depth": [3, None, 10, 20, 30],
    "max_features": randint(1, max(2, X_train_l1.shape[1])),
    "min_samples_leaf": randint(1, 9),
    "criterion": ["gini", "entropy", "log_loss"]
}
dt_rs = RandomizedSearchCV(
    DecisionTreeClassifier(random_state=42),
    param_distributions=param_dist_dt,
    n_iter=10, cv=5, scoring='accuracy', random_state=42, n_jobs=-1
)
dt_rs.fit(X_train_l1, y_train)
dt_best = dt_rs.best_estimator_
dt_pred_a = dt_best.predict(X_test_l1)
acc_dt_a = accuracy_score(y_test, dt_pred_a)

print("\n[Decision Tree] best params:", dt_rs.best_params_)
print(f"[Decision Tree] accuracy after tuning: {acc_dt_a:.4f}")

# -------------------------
# Summary table (like theirs)
# -------------------------
summary = pd.DataFrame({
    'Model': ['Random Forest','KNN','Decision Tree'],
    'Accuracy (no tuning)': [round(acc_rf_b*100,3), round(acc_knn_b*100,3), round(acc_dt_b*100,3)],
    'Accuracy (randomized search)': [round(acc_rf_a*100,3), round(acc_knn_a*100,3), round(acc_dt_a*100,3)],
})
print("\nSummary (%)")
print(summary)

# -------------------------
# Comparison chart
# -------------------------
barWidth = 0.25
fig = plt.subplots(figsize=(6, 4))

before_rs = summary['Accuracy (no tuning)'].values
after_rs = summary['Accuracy (randomized search)'].values

br1 = np.arange(len(before_rs))
br2 = br1 + barWidth

plt.bar(br1, before_rs, width=barWidth, edgecolor='grey', label='Before tuning')
plt.bar(br2, after_rs, width=barWidth, edgecolor='grey', label='After tuning')

plt.xlabel('Model')
plt.ylabel('Accuracy [%]')
plt.xticks(br1 + barWidth/2, summary['Model'])
plt.title("Performance comparison with randomized search")
plt.legend()
plt.ylim([min(min(before_rs),min(after_rs))-1, 100])
plt.tight_layout()
plt.show()

# -------------------------
# Save selected train/test (optional, same spirit)
# -------------------------
# X_train_l1.to_pickle("X_train_selected.pkl")
# X_test_l1.to_pickle("X_test_selected.pkl")

# Also print richer report for best model (RF here)
print("\nClassification report (RF, tuned):")
print(classification_report(y_test, rf_pred_a))
print("Confusion matrix (RF, tuned):")
print(confusion_matrix(y_test, rf_pred_a))
```

```python
# ============================================
# Model Selection (GPU-aware) for mushroom data
# ============================================
import numpy as np
import pandas as pd
import warnings

warnings.filterwarnings("ignore")

# ---------- Config ----------
RANDOM_STATE = 42
HAVE_TORCH = False
try:
    import torch
    HAVE_TORCH = True
except Exception:
    pass

# ---------- 0) Prepare data ----------
# Expect df already in memory with a 'class' column ('e'/'p' or 0/1)
df = df.copy()

# Target y
y_raw = df['class']
if pd.api.types.is_categorical_dtype(y_raw) or y_raw.dtype.kind in "OUS":
    y = y_raw.astype(str).map({'e': 0, 'p': 1}).astype(int)
else:
    y = y_raw.astype(int)

# Features X (no manual encoding here)
X = df.drop(columns='class')

# Optional: use your previously selected features if provided
try:
    feature_list = list(selected_features)
    # keep only features that exist in X
    feature_list = [c for c in feature_list if c in X.columns]
    if feature_list:
        X = X[feature_list]
    else:
        feature_list = X.columns.tolist()
except NameError:
    feature_list = X.columns.tolist()

# Train / test split on RAW X (pipelines handle encoding)
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=RANDOM_STATE, stratify=y
)

# ---------- 1) Preprocessing (inside pipelines) ----------
from sklearn.compose import ColumnTransformer, make_column_selector as selector
from sklearn.preprocessing import OrdinalEncoder, StandardScaler

# Selectors for column types from raw X
cat_sel = selector(dtype_include=['object', 'category', 'string'])
num_sel = selector(dtype_include=['number'])

# Categorical encoder that tolerates unseen categories in test
pre_cat = OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1)

# ColumnTransformer: encode categoricals, passthrough numerics
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', pre_cat, cat_sel),
        ('num', 'passthrough', num_sel),
    ],
    remainder='drop'
)

# ---------- 2) Utilities ----------
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

def evaluate_clf(name, model, Xtr, ytr, Xte, yte, proba_fn='predict_proba'):
    model.fit(Xtr, ytr)
    y_pred = model.predict(Xte)

    # Probabilities (or decision_function fallback) for ROC-AUC
    if hasattr(model, proba_fn):
        y_proba = getattr(model, proba_fn)(Xte)
        if isinstance(y_proba, np.ndarray) and y_proba.ndim == 2:
            y_proba = y_proba[:, 1]
    elif hasattr(model, 'decision_function'):
        s = model.decision_function(Xte)
        s_min, s_max = s.min(), s.max()
        y_proba = (s - s_min) / (s_max - s_min + 1e-12)
    else:
        y_proba = y_pred.astype(float)

    # Robust metrics even if a model predicts one class
    prec = precision_score(yte, y_pred, zero_division=0)
    rec  = recall_score(yte, y_pred, zero_division=0)
    f1   = f1_score(yte, y_pred, zero_division=0)
    roc  = roc_auc_score(yte, y_proba)

    return {
        "Model": name,
        "Accuracy": round(accuracy_score(yte, y_pred), 4),
        "Precision": round(prec, 4),
        "Recall": round(rec, 4),
        "F1": round(f1, 4),
        "ROC-AUC": round(roc, 4),
        "estimator": model
    }

results = []

```

```python

# ---------- 3) Models ----------
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neural_network import MLPClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

# For LR/SVM/MLP: encode -> scale -> model
scale_after_encode = StandardScaler(with_mean=True, with_std=True)

lr_pipe = Pipeline([
    ('prep', preprocessor),
    ('scaler', scale_after_encode),
    ('clf', LogisticRegression(max_iter=200, solver="liblinear", random_state=RANDOM_STATE))
])

svm_pipe = Pipeline([
    ('prep', preprocessor),
    ('scaler', scale_after_encode),
    ('clf', SVC(kernel="rbf", probability=True, random_state=RANDOM_STATE))
])

mlp_pipe = Pipeline([
    ('prep', preprocessor),
    ('scaler', scale_after_encode),
    ('clf', MLPClassifier(hidden_layer_sizes=(64, 32), max_iter=500, random_state=RANDOM_STATE))
])

# Trees don’t need scaling: encode -> model
rf_pipe = Pipeline([
    ('prep', preprocessor),
    ('clf', RandomForestClassifier(
        n_estimators=300, max_depth=30, random_state=RANDOM_STATE, n_jobs=-1
    ))
])
```

```python

# ---------- 3b) XGBoost with safe GPU→CPU fallback ----------
from xgboost import XGBClassifier
from xgboost.core import XGBoostError

xgb_params = dict(
    n_estimators=500, learning_rate=0.05, max_depth=6,
    subsample=0.8, colsample_bytree=0.8,
    random_state=RANDOM_STATE, eval_metric='logloss', use_label_encoder=False
)

# Always prepare a CPU pipe
xgb_cpu = XGBClassifier(tree_method="hist", predictor="auto", **xgb_params)
xgb_cpu_pipe = Pipeline([
    ('prep', preprocessor),
    ('clf', xgb_cpu)
])

# Try a GPU pipe, but fall back on failure at fit time
xgb_gpu = XGBClassifier(tree_method="gpu_hist", predictor="gpu_predictor", **xgb_params)
xgb_gpu_pipe = Pipeline([
    ('prep', preprocessor),
    ('clf', xgb_gpu)
])
```

```python

# ---- Evaluate with fallback ----
try:
    results.append(evaluate_clf("XGBoost (GPU)", xgb_gpu_pipe, X_train, y_train, X_test, y_test))
except XGBoostError as e:
    # Common case: "XGBoost version not compiled with GPU support."
    results.append(evaluate_clf("XGBoost (CPU)", xgb_cpu_pipe, X_train, y_train, X_test, y_test))

# ---------- 4) Evaluate classical models ----------
results.append(evaluate_clf("Logistic Regression", lr_pipe, X_train, y_train, X_test, y_test))
results.append(evaluate_clf("SVM (RBF)",           svm_pipe, X_train, y_train, X_test, y_test))
results.append(evaluate_clf("Random Forest",       rf_pipe, X_train, y_train, X_test, y_test))
```

```python

# ---------- 5) Optional: PyTorch NN (GPU if available) ----------
if HAVE_TORCH:
    # Fit the preprocessor on train only, then transform to numeric arrays
    # (Clone to avoid interfering with pipelines above)
    from sklearn.base import clone
    prep_for_nn = clone(preprocessor)
    Xtr_enc = prep_for_nn.fit_transform(X_train)
    Xte_enc = prep_for_nn.transform(X_test)

    # Scale features for NN
    scaler_nn = StandardScaler()
    Xtr_s = scaler_nn.fit_transform(Xtr_enc.astype(np.float32))
    Xte_s = scaler_nn.transform(Xte_enc.astype(np.float32))

    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    torch.manual_seed(RANDOM_STATE)

    Xtr_t = torch.tensor(Xtr_s, dtype=torch.float32).to(device)
    ytr_t = torch.tensor(y_train.values.reshape(-1, 1), dtype=torch.float32).to(device)
    Xte_t = torch.tensor(Xte_s, dtype=torch.float32).to(device)
    yte_t = torch.tensor(y_test.values.reshape(-1, 1), dtype=torch.float32).to(device)

    model_nn = torch.nn.Sequential(
        torch.nn.Linear(Xtr_t.shape[1], 64),
        torch.nn.ReLU(),
        torch.nn.Linear(64, 32),
        torch.nn.ReLU(),
        torch.nn.Linear(32, 1),
        torch.nn.Sigmoid()
    ).to(device)

    criterion = torch.nn.BCELoss()
    optimizer = torch.optim.Adam(model_nn.parameters(), lr=1e-3)

    model_nn.train()
    EPOCHS = 15
    BATCH = 1024
    for epoch in range(EPOCHS):
        perm = torch.randperm(Xtr_t.size(0), device=device)
        for i in range(0, Xtr_t.size(0), BATCH):
            idx = perm[i:i+BATCH]
            xb, yb = Xtr_t[idx], ytr_t[idx]
            optimizer.zero_grad()
            out = model_nn(xb)
            loss = criterion(out, yb)
            loss.backward()
            optimizer.step()

    model_nn.eval()
    with torch.no_grad():
        proba = model_nn(Xte_t).detach().cpu().numpy().ravel()
        pred = (proba >= 0.5).astype(int)

    results.append({
        "Model": f"Neural Net (PyTorch, {'GPU' if device.type=='cuda' else 'CPU'})",
        "Accuracy": round(accuracy_score(y_test, pred), 4),
        "Precision": round(precision_score(y_test, pred, zero_division=0), 4),
        "Recall": round(recall_score(y_test, pred, zero_division=0), 4),
        "F1": round(f1_score(y_test, pred, zero_division=0), 4),
        "ROC-AUC": round(roc_auc_score(y_test, proba), 4),
        "estimator": ("pytorch", model_nn, prep_for_nn, scaler_nn, device)
    })
else:
    # sklearn MLP fallback (already evaluated as mlp_pipe if you want it)
    pass
    # results.append(evaluate_clf("Neural Net (sklearn, CPU)", mlp_pipe, X_train, y_train, X_test, y_test))
```
